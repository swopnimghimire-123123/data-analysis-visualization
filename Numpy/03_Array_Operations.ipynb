### Array Operations

```python
import numpy as np
```

```python
a1 = np.array([1,2,3,4,5])
a2 = np.array([6,7,8,9,10])
```

```python
# when you want to perform some operations then the length of the array must be same

print(a1 + a2)
print(a1 - a2)
print(a1 * a2)
print(a1 / a2)
print(a1 // a2)
print(a1 % a2)
print(a1 ** a2)
```

```python
l = [10,20,30,30]
arr = np.array(l)
```

```python
arr + 10
```

```python
arr * 2
```

```python
arr2 = np.arange(1,26).reshape(5,5)
arr2
```

```python
arr2 * 2
```

```python
arr2 + 20
```

### Deep and shallow copy

```python
a = np.arange(1,21)
a
```

```python
slice = a[:5]
print(slice)
slice = slice * 10
print(slice)
```

```python
a
```

```python
a[4:6] * 10
```

```python
b = a.copy()
print("b[0] = ",b[0])
b[0] = 41
print(b)
```

### Advance Array Manupulation

```python
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
```

```python
np.vstack((a,b))
```

```python
np.hstack((a,b))
```
