### General Questions

```python
import numpy as np

# columns : [Age, Math Marks, Science Marks]
data = np.array(([18,56,98],
                 [10,89,78],
                 [15,66,78],
                 [10,98,98],
                 [15,66,78],
                 [18,56,35]
                 ))
```

```python
# Get the shape of the matrix
data.shape
```

```python
# Find the average of the first column
np.mean(data[:,0])
```

```python
# Extract math marks of all students
data[:,1]
```

```python
# Find tha maximum science marks
np.max(data[:,2])
```

```python
# Get details of the student who scored more than 90 marks in maths
data[data[:,1] > 90]
```

```python
# Increase math marks of all the students b 5
data[:,1] = data[:,1] + 5
data
```

```python
# print how many students are younger than 19
print(data[data[:,0] < 19].shape[0])
print(data[data[:,0] < 19])
```

```python
# Calculate the average marks of each subject
np.mean(data[:,1:],axis = 0)
```

```python
# Get data of students who have scored atleast 80 in both subject
math_condition = data[:, 1] >= 80
science_condition = data[:, 2] >= 80
display(data[math_condition & science_condition])
```

```python
# Replace all science marks < 75 with 0
data[:,2][data[:,2] < 75] = 0
print(data)
```
