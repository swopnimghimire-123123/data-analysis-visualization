```python
import numpy as np
import pandas as pd
```

###**GroupBy**

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
# Group by catagory and calculate the sum of sales
category_sales = df.groupby('Category')['Sales'].sum()
category_sales
```

```python
# Group by store and calculate the sum of sales
store_sales = df.groupby('Store')['Sales'].sum()
store_sales
```

```python
# Gropu by multiple columns
# Group by catagory and store
category_store_sales = df.groupby(['Category', 'Store'])['Sales'].sum()
category_store_sales
```

```python
# Swapping the values
category_store_sales = df.groupby(['Store', 'Category'])['Sales'].sum()
category_store_sales
```
