### Numpy indexing and slicing of vectors

```python
import numpy as np
```

```python
arr = np.arange(11,21)
arr
```

```python
arr[4]
```

```python
arr[1:5]
```

```python
arr[:5]
```

```python
arr[3:]
```

```python
arr[3::2]
```

```python
arr[::-1]
```

### Numpy Indexing and slicing of matrix

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0]
```

```python
arr[0,4]
```

```python
arr[5]
```

```python
arr[5,3]
```

```python
slice = arr[0:2,1:3]
slice
```

```python
slice = arr[3:6,3:5]
slice
```

```python
print(arr)
print(arr[:,2])
```

### Boolean Indexing

```python
arr = np.arange(11,21)
arr
```

```python
# Create a boolean index that is True for even numbers
bool_index = arr % 2 == 0

# Use the boolean index to select even numbers from the array
arr = arr[bool_index]
```

```python
# Display the array containing only even numbers
arr
```
