```python
import seaborn as sns
import matplotlib.pyplot as plt
```

### **Heat Map**

```python
flights = sns.load_dataset("flights")
```

```python
flights.head()
```

```python
tips = sns.load_dataset("tips")
```

```python
tips.head()
```

```python
tipscorr = tips[['total_bill', 'tip', 'size']]
```

```python
tipscorr
```

```python
tipscorr = tipscorr.corr()
```

```python
display(tipscorr)
```

```python
sns.heatmap(tipscorr.corr(),annot= True)
```

```python
sns.heatmap(tipscorr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap of Tips Data')
plt.show()
```

### Cluster Map

```python
sns.clustermap(tipscorr)
```
