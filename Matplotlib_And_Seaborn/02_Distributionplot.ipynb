```python
import seaborn as sns
import matplotlib.pyplot as plt
```

**We use distribution plots to visualize the distribution of quantitative data.**

Let's discuss some plots that allow us to visualize the distribution of a data set. These plots are:

- distplot/histplot
- jointplot
- pairplot
- rugplot

```python
df = sns.load_dataset('tips')
```



1.  **`plt.subplot(1,2,1)`**: This command sets up a grid for plotting. It specifies that we want a grid of 1 row and 2 columns, and the next plot will be placed in the first position (the first column of the first row).
2.  **`sns.histplot(df['total_bill'], bins=20, kde=True)`**: This uses Seaborn's `histplot` function to create a histogram of the `total_bill` column.
    *   `df['total_bill']`: Specifies the data to plot.
    *   `bins=20`: Divides the data into 20 bins for the histogram.
    *   `kde=True`: Adds a Kernel Density Estimate (KDE) curve on top of the histogram, which provides a smoothed representation of the distribution.
3.  **`plt.subplot(1,2,2)`**: This command switches to the second position in the 1x2 grid (the second column of the first row) for the next plot.
4.  **`sns.histplot(df['tip'], bins=20, kde=True)`**: Similar to the previous step, this creates a histogram of the `tip` column with 20 bins and a KDE curve.

In essence, this code generates two plots within a single figure: one showing the distribution of the `total_bill` and the other showing the distribution of the `tip` amount. The histograms show the frequency of values within different ranges, and the KDE curves provide a smoothed estimate of the probability density function.

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins = 20, kde=True)
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins = 20,kde = True)
```


- **`sns.jointplot()`**: This is the core function from the Seaborn library used to create a joint plot. A joint plot displays the relationship between two variables and also shows the individual distributions of each variable on the margins.
- **`x = 'total_bill'`**: This specifies that the 'total_bill' column from the DataFrame will be plotted on the x-axis.
- **`y = 'tip'`**: This specifies that the 'tip' column from the DataFrame will be plotted on the y-axis.
- **`data = df`**: This indicates that the data for the plot should come from the DataFrame named `df`.
- **`kind = 'scatter'`**: This argument determines the type of plot to be displayed in the central area of the joint plot. In this case, `'scatter'` creates a scatter plot, where each point represents a single observation with its corresponding 'total_bill' and 'tip' values.

In summary, this code generates a joint plot that visualizes the relationship between the 'total_bill' and 'tip' columns using a scatter plot in the center, and also includes histograms or kernel density estimates on the top and right margins to show the individual distributions of 'total_bill' and 'tip', respectively.

```python
sns.jointplot(x = 'total_bill', y = 'tip', data = df,kind = 'scatter')
```

```python
sns.jointplot(x="total_bill", y="tip", data=df, kind="reg")
```

The code `sns.pairplot(df)` generates a grid of plots that visualize the relationships between all pairs of numerical columns in your DataFrame `df`.

Here's what you can expect:

-   **Diagonal Plots**: On the diagonal of the grid, you will see the distribution of each individual numerical column. By default, this is a histogram, but it can be changed.
-   **Off-Diagonal Plots**: In the off-diagonal plots, you will see scatter plots showing the relationship between every possible pair of numerical columns. This allows you to quickly see correlations or patterns between different variables.

Essentially, `pairplot` is a convenient way to get a quick overview of the relationships and distributions of all numerical variables in your dataset.

```python
sns.pairplot(df)
```

```python
sns.pairplot(df,hue="sex")
```

```python
sns.pairplot(df, hue="size", palette="coolwarm")
```

```python
sns.pairplot(df, hue="size", palette="rainbow")
```
