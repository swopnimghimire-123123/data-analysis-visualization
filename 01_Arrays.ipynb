### Basic array creation

```python
import numpy as np
# Import the numpy library, commonly used for numerical operations.
```

```python
# Create a NumPy array from a Python list
arr = np.array([1,2,3,4,5])
arr
```

```python
# Create a Python list with mixed data types
a = [1,2,3,"Hello"]
a
```

```python
# Convert a Python list with mixed data types to a NumPy array
# Note that NumPy converts all elements to strings in this case
np.array(a)
```

```python
# Create a nested Python list representing a 2D structure
l = [[1,2,3],[4,5,6],[7,8,9]]
l
```

```python
# Convert a nested Python list to a 2D NumPy array
np.array(l)
```

### Array generation function

```python
# Create a NumPy array with a range of values using np.arange()
# The array starts at 1, ends before 11, and steps by 2
arr = np.arange(1,11,2)
arr
```

```python
# Create a 6x6 NumPy array filled with zeros
arr = np.zeros((6,6))
arr
```

```python
# Create a 2x5 NumPy array filled with zeros
arr = np.zeros((2,5))
arr
```

```python
# Create a 6x6 NumPy array filled with ones
arr = np.ones((6,6))
arr
```

```python
# Create a NumPy array with values evenly spaced between 1 and 5
arr = np.linspace(1,5,2)
print(arr)
# Create another NumPy array with values evenly spaced between 1 and 20
brr = np.linspace(1,20,3)
print(brr)
```

### Random Generation Function

```python
# Generate an array of 10 random numbers from a uniform distribution between 0 and 1
np.random.rand(10)
```

```python
# Generate an array of 10 random numbers from a standard normal distribution (mean 0, variance 1)
np.random.randn(10)
```

```python
# Generate an array of 9 random integers between 10 (inclusive) and 20 (exclusive)
np.random.randint(10,20,9)
```

```python
# Get the data type of the elements in the array
crr.dtype
```

### Array Methods

```python
# Display the array again for reference
arr
```

```python
# Find the minimum value in the array
arr.min()
```

```python
# Find the maximum value in the array
arr.max()
```

```python
# Calculate the sum of all elements in the array
arr.sum()
```

```python
# Calculate the sum of elements along axis 0 (columns) for the array 'arr'
# Note: This likely refers to a previous 'arr' which was a 2D array.
np.sum(arr,axis = 0)
```

```python
# Calculate the sum of elements along axis 1 (rows) for the array 'arr'
# Note: This likely refers to a previous 'arr' which was a 2D array.
np.sum(arr,axis = 1)
```

```python
# Calculate the mean of the elements in the array 'arr'
# Note: This likely refers to a previous 'arr' which was a 2D array.
arr.mean()
```

```python
# Calculate the standard deviation of the elements in the array 'arr'
# Note: This likely refers to a previous 'arr' which was a 2D array.
arr.std()
```

```python
# Get the index of the maximum value in the flattened array 'arr'
# Note: This likely refers to a previous 'arr' which was a 2D array.
arr.argmax()
```

```python
# Get the index of the minimum value in the flattened array 'arr'
# Note: This likely refers to a previous 'arr' which was a 2D array.
arr.argmin()
```

### Reshape and resizing

```python
# Create a NumPy array with a range of values from 1 to 30
arr = np.arange(1,31)
arr
```

```python
# Reshape the array 'arr' into a 6x5 matrix
arr = arr.reshape(6,5)
```
