```python
import pandas as pd
import numpy as np
```

### **Finding Missing Data**

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

###**Removing Missing Data**

```python
df
```

```python
df.dropna()
```

```python
df
```

```python
df.dropna(thresh=1)
# thresh = minimum number of non-NA values required to keep the row/column.
# If a row/column has fewer non-NA values than thresh, it gets dropped.
```

###**Filling the missing Data**

```python
df
```

```python
df.fillna(0)
```

```python
values = {'A': 0, 'B': 1, 'C': 2, 'D': 3}
df.fillna(value=values)
```

```python
df
```

```python
df.fillna(df.mean())
# just look at the column D the sum of 1 and 5 is 6 and it's mean is 3
# so all the null values are repalced by the mean
```
